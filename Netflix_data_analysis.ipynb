```python
## Movies Released in the Year 2000
## Filter and analyze movies released during the year 2000.
df['Year'] = df['Date_N'].dt.year
```

```python
## Netflix Content Produced in India
## Identify TV Shows that were released in India.
df[(df['Category']=='TV Show') & (df['Country']=='India')]['Title']
```

```python
## Most Prolific Directors on Netflix
## Determine the top 10 directors with the highest number of titles on the platform.
df['Director'].value_counts().head(10)
```

```python
## Content Filtering by Genre and Country
## Analyze Movies categorized as Comedies and content originating from the United Kingdom.
df[((df['Category']=='Movie')&(df['Type']=='Comedies')) | (df['Country']=='United Kingdom')]
```

```python
## Tom Cruise Content Analysis
## Identify all Movies and TV Shows featuring Tom Cruise.
df[df['Cast'].str.contains('Tom Cruise')]
```

```python
## Netflix Rating Categories
## Explore and summarize the rating classifications used by Netflix.
df.Rating.unique()
```

```python
## TV-14 Content Analysis in Canada
## Determine the number of titles rated TV-14 in Canada.
df[(df['Category']=='Movie') & (df['Rating']=='TV-14') & (df['Country']=='Canada')].shape
```

```python
## R-Rated TV Shows Released After 2018
## Analyze TV Shows with an R rating added after 2018.
df[(df['Category']=='TV Show') & (df['Rating']=='R') & (df['Year']>2018)]
```
